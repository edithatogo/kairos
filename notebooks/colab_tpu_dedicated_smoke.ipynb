```python
import json
import platform

devices = jax.devices()
print(json.dumps({
    "platform": platform.platform(),
    "backend": jax.default_backend(),
    "jax_version": jax.__version__,
    "devices": [str(device) for device in devices],
}, indent=2))
assert any("TPU" in str(device).upper() for device in devices), f"No TPU device found: {devices}"
```

```python
@jax.jit
def step(x):
    return jnp.matmul(x, x) + 1.0

x = jnp.arange(16, dtype=jnp.float32).reshape(4, 4)
y = step(x)
print(y)
print(json.dumps({"status": "ok", "sum": float(jnp.sum(y))}))
```
