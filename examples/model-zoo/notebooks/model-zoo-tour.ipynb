# KairoECS Model Zoo Tour

This notebook is a documentation-first walkthrough. It explains how to read the model-zoo inventory, choose an example by maturity, and identify expected outputs before any runtime backend is required.

## 1. Pick by maturity

| Maturity | Use it for |
|---|---|
| `benchmark` | Benchmark or regression-oriented documentation. |
| `reference` | Recommended example shape for a paradigm. |
| `domain-preview` | Domain tutorial planning with bounded claims. |

Keep labels conservative until a command or checked artifact proves the stronger claim.
